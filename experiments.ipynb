```python
from pathlib import Path

from bootstrapagent.client import Client
from bootstrapagent.configs import CONFIG
from bootstrapagent.model_definitions import (
    init_embedding_model,
    init_transformer_model,
)

transformer_model = init_transformer_model(
    gpu_number=0,
    device=CONFIG.device,
    temperature=CONFIG.transformer_model.temperature,
    model_name=CONFIG.transformer_model.name,
)

embedding_model = init_embedding_model(
    gpu_number=0,
    device="gpu",
    padding=CONFIG.embedding_model.padding,
    dimensions=CONFIG.embedding_model.dimensions,
    model_name=CONFIG.embedding_model.name,
)

client = Client(
    transformer_model=transformer_model,
    embedding_model=embedding_model,
)
```

```python
primitive_dir_candidates = [
    Path("library/primtives"),
    Path("src/library/primtives"),
    Path("library/primitives"),
    Path("src/library/primitives"),
]

primitive_dir = next((path for path in primitive_dir_candidates if path.exists()), None)
if primitive_dir is None:
    raise FileNotFoundError("Could not find /library/primtives or /library/primitives")

function_paths = sorted(primitive_dir.glob("*.py"))
function_texts = [
    path.read_text(encoding="utf-8")
    for path in function_paths
    if path.is_file()
]

test_result = client.test(function_texts=function_texts)
test_result
```
